```python
def get_af2_pdb(uniprot):
    api_url = f"https://alphafold.ebi.ac.uk/api/prediction/{uniprot}"
    resp = requests.get(api_url)
    resp.raise_for_status()
    data = resp.json()
    if not data:
        raise ValueError(f"No AlphaFold model found for {uniprot}")

    pdb_url = data[0]["pdbUrl"]
    local_path = os.path.basename(pdb_url)

    if not os.path.exists(local_path):
        r = requests.get(pdb_url)
        r.raise_for_status()
        with open(local_path, "wb") as f:
            f.write(r.content)

    return local_path

def get_plddt(pdb_path):
    plddt = {}
    with open(pdb_path) as f:
        for line in f:
            if line.startswith("ATOM") and line[12:16].strip() == "CA":
                resnum = int(line[22:26])
                plddt[resnum] = float(line[60:66])
    return plddt

def get_pdb_sequence(pdb_path):
    three_to_one = {
        'ALA':'A','ARG':'R','ASN':'N','ASP':'D','CYS':'C','GLN':'Q','GLU':'E',
        'GLY':'G','HIS':'H','ILE':'I','LEU':'L','LYS':'K','MET':'M','PHE':'F',
        'PRO':'P','SER':'S','THR':'T','TRP':'W','TYR':'Y','VAL':'V',
    }
    seq = {}
    with open(pdb_path) as f:
        for line in f:
            if line.startswith("ATOM") and line[12:16].strip() == "CA":
                resnum = int(line[22:26])
                resname = line[17:20].strip()
                seq[resnum] = three_to_one.get(resname, "X")
    return seq

from difflib import SequenceMatcher

def align_residue_numbering(bmrb_seq, bmrb_first_resnum, pdb_seq_dict):

    pdb_resnums = sorted(pdb_seq_dict)
    pdb_seq = "".join(pdb_seq_dict[r] for r in pdb_resnums)

    matcher = SequenceMatcher(None, bmrb_seq, pdb_seq, autojunk=False)
    match = matcher.find_longest_match(0, len(bmrb_seq), 0, len(pdb_seq))

    if match.size < 10:   # too short to trust as a real alignment
        return None

    return {
        bmrb_first_resnum + match.a + i: pdb_resnums[match.b + i]
        for i in range(match.size)
    }
```

```python
from makeshift.rci import RCI

outputs = []
for _, row in df.iterrows():
    try:

        #
        cs = ms.ChemicalShifts.from_bmrb(row['bmrb'], keep_download=True)
        rci = RCI().calc(cs)

        #
        bmrb_seq = cs.entry.sequences()['Polymer_seq_one_letter_code'].iloc[0]
        bmrb_first_resnum = int(cs.data['Seq_ID'].min())

        #
        pdb_path = get_af2_pdb(row['uniprot'])
        plddt = get_plddt(pdb_path)
        pdb_seq = get_pdb_sequence(pdb_path)

        resnum_map = align_residue_numbering(bmrb_seq, bmrb_first_resnum, pdb_seq)
        if resnum_map is None:
            print(f"Could not align sequences for {row['bmrb']}")
            continue

        tmp = rci.results.copy()
        tmp['protein'] = row['bmrb']
        tmp['pdb_resnum'] = tmp['Seq_ID'].map(resnum_map)
        tmp['plddt'] = tmp['pdb_resnum'].map(plddt)

        outputs.append(tmp)

    except Exception as e:
        print(f"Error getting {row['bmrb']}: {e}")

outputs = pd.concat(outputs, ignore_index=True)
outputs = outputs.dropna(subset=['pdb_resnum'])
```
